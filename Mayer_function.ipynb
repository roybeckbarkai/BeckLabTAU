```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact

# Define the U function
def U(x):
    return 1 / x**12 - 1 / x**6

# Define the function to plot
def plot_function(T=10):
    x = np.linspace(0.2, 5, 500)
    y1 = U(x)
    y2 = np.exp(-U(x) / T)

    plt.figure(figsize=(8, 6))
    plt.plot(x, y1, label=r'$U(x)$')
    plt.plot(x, y2, label=r'$e^{-U(x)/T}$')
    plt.ylim(-1, 4)
    plt.xlabel('x')
    plt.ylabel('Function Value')
    plt.title(f'Plot for T = {T}')
    plt.legend()
    plt.grid()
    plt.show()

# Create interactive plot
interact(plot_function, T=(0.1, 3, 0.1))
```
